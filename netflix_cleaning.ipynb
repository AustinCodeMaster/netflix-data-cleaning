```python
df_clean = df.copy()
```

```python
mask = df_clean["duration"].isna() & df_clean["rating"].str.fullmatch(r"\d+ min", na=False)
```

```python
df_clean[mask]
```

```python
df_clean.loc[mask, "duration"] = df_clean.loc[mask, "rating"]
```

```python
df_clean.loc[mask, "rating"] = pd.NA
```

```python
df_clean.loc[mask, ["title", "rating", "duration"]]
```

```python
df_clean[["rating", "duration"]].isna().sum()
```

```python
df_clean["rating"].value_counts(dropna=False)
```

```python
df_clean["rating"] = df_clean["rating"].astype("string")
```

```python
df_clean["rating"].value_counts(dropna=False).sum()
```

```python
df_clean.groupby("type")["director"].apply(lambda column: column.isna().sum())
```

```python
df_clean.loc[df_clean["cast"].isna(), ["title", "type", "listed_in"]].head(10)
```

```python
df_clean.loc[df_clean["country"].isna(), ["title", "type", "country"]].head(10)
```

```python
df_clean.loc[df_clean["date_added"].isna(), ["title", "date_added", "release_year"]]
```

```python
df_clean["date_added"] = pd.to_datetime(
    df_clean["date_added"].str.strip(),
    format="%B %d, %Y"
)
```

```python
df_clean["date_added"].head()
```

```python
df_clean["date_added"].isna().sum()
```

```python
df_clean["duration"].str.split(" ", n=1, expand=True).head()
```

```python
df_clean[["duration_value", "duration_unit"]] = (
    df_clean["duration"].str.split(" ", n=1, expand=True)
)
```

```python
df_clean[["duration", "duration_value", "duration_unit"]].head()
```

```python
df_clean["duration_value"] = df_clean["duration_value"].astype("Int64")
```

```python
df_clean["duration_unit"] = df_clean["duration_unit"].replace({
    "Season": "season",
    "Seasons": "season"
})
```

```python
df_clean[["duration_value", "duration_unit"]].head()
```

```python
df_clean.duplicated().sum()
```

```python
df_clean.info()
```

```python
df_clean.to_csv("netflix_titles_cleaned.csv", index=False)
```
